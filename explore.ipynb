```python
# new cell in explore.ipynb or paste in a .py file
import sys
sys.stdout.reconfigure(encoding="utf-8")
from datasets import load_dataset
from collections import Counter

ds = load_dataset("arbml/ashaar")["train"]
lang_counts = Counter(ds["poem language type"])
print(f"Total rows: {len(ds):,}")
print(f"\nLanguage type distribution:")
for val, count in lang_counts.most_common(15):
    pct = count / len(ds) * 100
    print(f"  {str(val)!r:35s} {count:>7,}  ({pct:.1f}%)")
```
